# 01. Data Acquisition - COVID-19 Data Tracking System

**Objective**: Download the official COVID-19 dataset, load it into a pandas DataFrame, and perform initial exploration.

**Dataset Source**: Our World in Data (OWID) - Complete COVID-19 dataset  
This dataset contains daily records of confirmed cases, deaths, recoveries, vaccinations, testing, and many other metrics for 200+ countries.

```python
import os
import requests
import pandas as pd
import numpy as np
from pathlib import Path
from tqdm import tqdm

# Display settings
pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', 100)
pd.set_option('display.width', 1000)
pd.set_option('display.float_format', '{:.2f}'.format)

print("Libraries imported successfully!")
```

```python
# Project root (go up one level from notebooks/)
PROJECT_ROOT = Path.cwd().parent
RAW_DATA_DIR = PROJECT_ROOT / "data" / "raw"
RAW_DATA_DIR.mkdir(parents=True, exist_ok=True)

print(f"Project Root : {PROJECT_ROOT}")
print(f"Raw Data Dir : {RAW_DATA_DIR}")
```

```python
# Try the current official OWID catalog link
DATA_URL = "https://catalog.ourworldindata.org/garden/covid/latest/compact/compact.csv"
OUTPUT_FILE = RAW_DATA_DIR / "owid-covid-data.csv"

def download_file(url, output_path):
    """Download a file with progress bar"""
    try:
        response = requests.get(url, stream=True, timeout=30)
        response.raise_for_status()
        total_size = int(response.headers.get('content-length', 0))
        
        with open(output_path, 'wb') as file, tqdm(
            desc=output_path.name,
            total=total_size,
            unit='B',
            unit_scale=True,
            unit_divisor=1024,
        ) as bar:
            for data in response.iter_content(chunk_size=1024):
                size = file.write(data)
                bar.update(size)
        
        print(f"\n✅ Download completed: {output_path}")
        print(f"File size: {output_path.stat().st_size / (1024*1024):.2f} MB")
        return True
    except Exception as e:
        print(f"❌ Failed to download from {url}")
        print(f"Error: {e}")
        return False

# Download
if not OUTPUT_FILE.exists():
    print("Trying to download from Our World in Data catalog...")
    success = download_file(DATA_URL, OUTPUT_FILE)
    
    if not success:
        print("\n⚠️ Automatic download failed. Please use Solution 2 or 3 below.")
else:
    print(f"File already exists: {OUTPUT_FILE}")
```

```python
OUTPUT_FILE = RAW_DATA_DIR / "owid-covid-data.csv"
print(f"Looking for file at: {OUTPUT_FILE}")
print(f"File exists: {OUTPUT_FILE.exists()}")
```

```python
df = pd.read_csv(OUTPUT_FILE)

print("✅ Data loaded successfully!")
print(f"Shape of dataset: {df.shape}")
print(f"Number of rows   : {df.shape[0]:,}")
print(f"Number of columns: {df.shape[1]}")
```

```python
print("="*60)
print("BASIC DATASET INFORMATION")
print("="*60)

print("\n1. First 5 rows:")
display(df.head())

print("\n2. Column names:")
print(df.columns.tolist())

print("\n3. Data types:")
print(df.dtypes)

print("\n4. Missing values (top 20 columns with most missing data):")
missing = df.isnull().sum().sort_values(ascending=False)
print(missing[missing > 0].head(20))
```

```python
print("Shape:", df.shape)
print("\nAll columns:")
print(df.columns.tolist())

print("\nFirst 5 rows:")
display(df.head())
```

```python
print("="*60)
print("KEY STATISTICS")
print("="*60)

print(f"\n• Unique countries/locations : {df['country'].nunique()}")
print(f"• Date range                 : {df['date'].min()} → {df['date'].max()}")
print(f"• Total records              : {len(df):,}")

print("\n• Top 10 countries by number of records:")
print(df['country'].value_counts().head(10))

print("\n• Continents available:")
print(df['continent'].value_counts(dropna=False))
```

```python
# Important columns we will use heavily later
important_cols = [
    'code', 'continent', 'country', 'date',
    'total_cases', 'new_cases', 'total_deaths', 'new_deaths',
    'total_cases_per_million', 'total_deaths_per_million',
    'reproduction_rate', 'icu_patients', 'hosp_patients',
    'total_tests', 'new_tests', 'positive_rate',
    'total_vaccinations', 'people_vaccinated', 'people_fully_vaccinated',
    'stringency_index', 'population', 'median_age',
    'gdp_per_capita', 'life_expectancy'
]

print("Important columns present in the dataset:")
for col in important_cols:
    status = "✅" if col in df.columns else "❌"
    print(f"{status} {col}")
```
