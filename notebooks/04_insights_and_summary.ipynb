# 04. Key Insights & Analytical Summary
**COVID-19 Data Tracking System**

This notebook summarizes the most important findings from the analysis in a clear, decision-oriented way.

```python
import pandas as pd
import numpy as np
from pathlib import Path

PROJECT_ROOT = Path.cwd().parent
df = pd.read_csv(PROJECT_ROOT / "data" / "processed" / "covid_cleaned.csv", parse_dates=['date'])

latest = df.sort_values('date').groupby('country').tail(1).copy()

print("Data loaded successfully")
print(f"Countries analyzed: {latest['country'].nunique()}")
```

```python
print("="*70)
print("GLOBAL OVERVIEW (Latest Available Data)")
print("="*70)

total_cases = latest['total_cases'].sum()
total_deaths = latest['total_deaths'].sum()
avg_cfr = (total_deaths / total_cases) * 100 if total_cases > 0 else 0

print(f"Total Confirmed Cases worldwide     : {total_cases:,.0f}")
print(f"Total Deaths worldwide              : {total_deaths:,.0f}")
print(f"Global Case Fatality Rate (CFR)     : {avg_cfr:.2f}%")
print(f"Number of countries analyzed        : {latest['country'].nunique()}")
```

```python
summary_table = latest[[
    'country', 'continent', 'total_cases', 'total_deaths', 
    'case_fatality_rate', 'fully_vaccinated_rate', 'population'
]].copy()

summary_table = summary_table.sort_values('total_cases', ascending=False).head(15)
summary_table = summary_table.reset_index(drop=True)
summary_table.index = summary_table.index + 1

# Format numbers
summary_table['total_cases'] = summary_table['total_cases'].apply(lambda x: f"{x:,.0f}")
summary_table['total_deaths'] = summary_table['total_deaths'].apply(lambda x: f"{x:,.0f}")
summary_table['case_fatality_rate'] = summary_table['case_fatality_rate'].apply(lambda x: f"{x:.2f}%")
summary_table['fully_vaccinated_rate'] = summary_table['fully_vaccinated_rate'].apply(lambda x: f"{x:.1f}%")
summary_table['population'] = summary_table['population'].apply(lambda x: f"{x:,.0f}")

print("Top 15 Most Affected Countries:")
display(summary_table)
```

## Key Analytical Insights

### 1. Outbreak Progression
- The pandemic showed multiple distinct waves globally.
- The highest peaks in daily new cases and deaths occurred during major variant-driven waves (especially Delta and Omicron periods).
- 7-day rolling averages clearly reveal the true trend by smoothing daily reporting noise.

### 2. Mortality Patterns
- Case Fatality Rate (CFR) varies significantly across countries.
- Countries with older populations and weaker healthcare systems generally showed higher mortality.
- Global CFR settled at a relatively low level in later stages due to vaccination and better treatment protocols.

### 3. Country-level Comparisons
- A small number of countries accounted for a very large share of total cases and deaths.
- High population countries (India, USA, Brazil, etc.) dominate absolute numbers.
- When adjusted for population (per million metrics), the ranking of severity changes significantly.

### 4. Vaccination Impact
- Countries that achieved high fully-vaccinated rates generally experienced lower death rates in later waves.
- Vaccination coverage shows strong variation between high-income and low-income countries.

### 5. Continental Differences
- Europe and North America recorded high absolute cases early in the pandemic.
- Asia and South America experienced severe later waves.
- Africa reported relatively lower cases, likely influenced by younger demographics and testing capacity differences.

## Decision-Oriented Summary

**What the data tells us:**

1. **Early detection & testing capacity** mattered a lot — countries with better testing systems reported cases earlier and could respond faster.
2. **Healthcare system strength** and population age structure strongly influenced mortality outcomes.
3. **Vaccination campaigns** significantly reduced the severity of later waves.
4. **Public health measures** (stringency index) had varying effectiveness depending on timing and compliance.
5. Absolute numbers can be misleading — **per-million metrics** give a fairer comparison between countries.

**Recommendations based on analysis:**
- Maintain strong surveillance systems even after the peak of a pandemic.
- Prioritize vaccination of vulnerable populations.
- Invest in healthcare capacity (ICU beds, hospital infrastructure).
- Use data-driven communication to guide public behavior.
